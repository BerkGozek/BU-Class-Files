# while

Our programs so far have just run to the end and stopped.  A while loop allows a program to keep going by circling back to an earlier point in the program.

```python
string = input('Enter a number: ')
while string != 'stop':
    print(string + ' squared is ' + str(int(string) ** 2))
    string = input('Enter a number: ')
print('Done.')
```

* **When the line with while is reached, the condition is checked**.  If it's True, we proceed to execute the while's indented block.
* If the condition was True, **the whole block is executed**, even if variables involved in the condition change.
* When the end of the indented block is reached, **execution jumps back to the condition check.**
* When the condition is checked and False, **execution jumps to after the while block.**

```python
# Counting from 0 to 20
counter = 0
while counter < 21:
    print(counter)
    counter = counter + 1
# What do you think the value of counter is at the end?
print(counter)
```

* The while code is called a **loop** because execution loops back to the condition at the beginning of the block.
* Each pass through the while block's code is called an **iteration**.  To **iterate** is to pass through the block multiple times.

```python
counter = 1
print('We will now iterate three times...')
while counter < 4:
    print('Iteration ' + str(counter))
    counter = counter + 1
```

With while, we can allow the user to keep entering more data until they're done.

```python
total = 0
count = 0
value_str = input('Enter a number, or "done" if done: ')
while value_str != 'done':
    count = count + 1
    value_int = int(value_str)
    total = total + value_int
    value_str = input('Enter a number, or "done" if done: ')
if count > 0:
    print('The average is ' + str(total/count))
```

On a slight (but important) tangent, there is a very commonly used shorthand we could use here.  "variablename += 1" is short for "variable = variable + 1."  It's very common to increase a variable value by a fixed amount.

```python
total = 0
count = 0
value_str = input('Enter a number, or "done" if done: ')
while value_str != 'done':
    count += 1
    value_int = int(value_str)
    total += value_int
    value_str = input('Enter a number, or "done" if done: ')
if count > 0:
    print('The average is ' + str(total/count))
```

Programs can run forever if the "while" condition is never False.  Maybe that's what you want; maybe it's not?

```python
# Desirable looping forever?
while(True):
    input('Enter any input to get a compliment: ')
    print('That is so clever of you!')
```

### Lists

In the programs above, we used running totals to get around the fact that we had only a finite number of variables to work with.  What if we wanted to remember all the specific values the user entered?

A list is a **data structure** that holds multiple pieces of data.

```python
my_list = ['duck', 'duck', 'goose']  # A list with 3 items
print(my_list[0])
print(my_list[1])
print(my_list[2])
```

<img src=" ">

Using that bracket notation ("**array notation**") we can treat each place in the list as a variable, including for purposes of assignment.

```python
my_list = ['duck', 'duck', 'goose']
my_list[2] = 'bear'
print(my_list)
```

Lists are **objects**, bundles of data with associated functions called **methods**.  Methods are very much like other functions but are called differently - we name the object, add a period, and then call the function.

append() is a list method - it takes an item as an argument and adds it to the end of the list.  It doesn't evaluate to anything - it just **destructively** modifies the list (changes its stored values).

```python
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)  # my_list has changed...
```

```python
# But the append call doesn't *evaluate* to anything
print(my_list.append(5))
print(my_list)
```

Often, we want to create a new **empty list** and then grow it by adding items to it, one at a time.  The empty list is denoted with two square brackets with nothing in between, []. This piece of code also introduces the string method, .lower(), which converts all the characteres in a string to lowercase letters.

```python
shopping_list = []
item = input('Add an item to the shopping list (or "done"): ')
while item.lower() != 'done':
    shopping_list.append(item)
    item = input('Add an item to the shopping list (or "done"): ')
print('Okay, so that was: ')
print(shopping_list)
```

Lists have some similarity to strings, which Python treats sort of like lists of letters.  Like strings, lists can have their contents concatenated with +.

```python
[1, 2, 3] + [4, 5, 6]
```

The function len() (short for "length") works on either strings or lists, and returns the number of characters in the string or the number of items in the list.

```python
print(len('Hello'))
print(len([1, 2, 3]))
```

If the first item in a list or string has index 0 instead of 1, then that shifts the index of everything else as well; the second item has index 1, the third item has index 2, and so on down to the last item, which has index len(L) - 1.  Trying to index the list at index len(L) results in an error.

```python
my_items = ['eggs', 'flour', 'milk']
print(len(my_items), 'items')
print(my_items[2])
print(my_items[len(my_items)-1])
```

There are several convenient methods built in to the list object besides append().  One is sort().  This conveniently sorts the items in the list from low to high, assuming Python knows how to compare them with each other.  If it can use < on the data, then the list method can sort() them.

```python
# Diameters in order from Sun:  Mercury, Venus, Earth, Mars... to Pluto, honorary planet
planet_diameter_km = [4879, 12104, 12756, 6792, 142984, 120536, 51118, 49528, 2377]
planet_diameter_km.sort()
planet_diameter_km
```

Like append(), sort() modifies the list but returns None.  This technically makes the sorting of sort() a **side effect**, just like the printing of print(), since they are effects unrelated to what they evaluate to (None).  It may not be the best name for this phenomenon.

There are more list methods available - see the documentation at https://docs.python.org/3/tutorial/datastructures.html.

Last, here's a weird gotcha that can cause bugs if you're unaware of it.

When you assign a list to a variable, you are actually assigning the list's **address in memory** to that variable.  Evaluating the variable seems to produce the list itself, but Python actually had to look up the variable's address to produce that value. We can think of the variable as holding a "pointer" to its data.


<img src=" ">

When a variable pointing to a list is assigned to another variable, the address is copied, and both variables will be holding that same address.  So if something like a sort() changes the list, both variables will see the change.

```python
my_list1 = [3, 2, 1]
my_list2 = my_list1
my_list1.sort()
print(my_list1)
print(my_list2)
```

If this is undesirable and you want to create a new copy of the list that changes separately, my_list.copy() returns a new copy of the list.

```python
my_list1 = [3, 2, 1]
my_list2 = my_list1.copy()
my_list1.sort()
print(my_list1)
print(my_list2)
```

## Review program:  Honor roll

In this program, the user is entering names that are checked against a hardcoded list of honors students.  Two lists are built up, one of entered names that are honors, the other a list of students that weren't on the list.  These are both reported in the end.  Lists are making the task of checking the honor roll easier, and they also mean we can remember arbitrarily long lists of names.

This example also introduces the in keyword; "a in L" evaluates to True if *L* contains *a* as one of its items.

```python
honors = ['Albert', 'Berenice', 'Chen', 'Dominique']
mentioned_honors = []
nonhonors = []
student = input('Enter a name (or "done"): ')
while (student != 'done'):
    if student in honors:
        print('Honors!')
        mentioned_honors.append(student)
    else:
        print('Not honors...')
        nonhonors.append(student)
    student = input('Enter a name (or "done"): ')
print('Honors mentioned: ' + str(mentioned_honors))
print('Nonhonors mentioned: ' + str(nonhonors))
```
