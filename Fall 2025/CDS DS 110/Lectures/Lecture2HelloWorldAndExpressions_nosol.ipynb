# Analyzing a first program:  Hello World

```python
print('Hello, world!')
```

* The **interpreter** is the program that interprets and executes our code.  It is run when we run the code box.

* A **statement** is a single line of a Python program - an instruction to the interpreter.  The "Hello, World" program is one statement long.

* Given multiple statements, the interpreter executes them *one at a time from top to bottom*.

```python
print('Hello 1')
print('Hello 2')
print('Hello 3')
```

* A piece of text that appears in quotation marks, like 'Hello, world!' or "Hello, world!", is called a **string**.

* Strings are treated as literal text by the interpreter instead of code to interpret.  Leave out the quotation marks, and you'll get an error unless the text happens to be meaningful code.

```python
print('Hello, world!')
```

```python
print("Hello, world!")
```

The text that appears after a # in code is called a **comment**.  It is ignored by the interpreter, but helps out other humans, especially if you're doing something unusual.

```python
print(Hello, world!) # Intentionally creates an error!
```

* An **expression** is a snippet of code that evaluates to something.

* Sample expressions include strings, numbers, or numbers that are combined using arithmetic **operators** such as +, -, *, /.

    * When expressions are acted on by operators, they're called **operands**.

* print() will accept any expression as its input.

```python
print(1)         # Technically an expression
print(1+2)       # Two operands and an operator make an expression
print(10*(10+1)) # The expression (10+1) acting as an operand
```

When evaluating expressions, there is an **order of operations**.  With the arithmetic we have so far, the rules are simple:

* When a value of an expression within parentheses is needed, evaluate it first.
* Multiplication and division come before addition and subtraction.
* Otherwise, evaluate from left to right.

**We will learn more operators, each with their own precedence.**

```python
print(3 + 8 / 2)  # What do you predict?
```

```python
print(4 * 2 + 3 + 5 * 2)  # And this one?
```

* print() is a **function** - it takes zero or more comma-separated inputs called **arguments**, and behaves differently depending on those inputs

    * In the case of print, the arguments are printed on the same line, separated by spaces.

```python
print('Hello', 'world', '!')
```

* Functions besides print() generally evaluate to something when called - they're expressions
* We say a function **returns** its value
    * For example, max() returns the largest of its arguments

```python
print(max(2,5,7))
```

```python
print(max(2,7) + max(3,9)) # Using function calls as operands
```

Functions are treated similarly to expressions in parentheses when it comes to order of evaluation.  So in the preceding example, when + needs values on either side, first the left max gets called, then the right max gets called, and then the addition takes place.

```python
print(max(2,7) + max(3,9)) # Calc 7, calc 9, then add
```

Incidentally, in notebooks print() is less necessary because **notebooks print what the last line in the code box evaluates to.**  (Unless that value is the special value **None**.)

```python
1
2
3
```

```python
max(2,7)
```

```python
None
```

**print() evaluates to None**, not what it prints.  So you can't meaningfully use it in an expression.

```python
print(2) + 2
```

These are the four most important **types** of data in Python.

* **int**:  Integer; a positive or negative number with no fractional/decimal component; computer represents it exactly

* **float**:  Floating point number; a number that can contain a decimal part or fraction; imprecise beyond a few significant figures

* **str**:  String

* **bool**:  Boolean value:  True or False

```python
print(type(-100)) # int
print(type(10.1)) # float
print(type('A'))  # str
print(type(True)) # bool
```

You can tell the difference between a string, int, and float because the string will have quotation marks, and the float will have at least one number after the decimal point.

```python
print(type('10')) # str
print(type(10))   # int
print(type(10.0)) # float
print(type(True)) # bool
```

The imprecision we saw earlier is caused by the fact that computers represent numbers in **binary**, bits that indicate which powers of two to sum.  Powers of 2 can sum to any integer, but some fractions, like 1/10, can only be represented as infinite sums of fractional powers of two (1/2, 1/4, ...), so their representation is an approximation.  (Exact computations can use the Decimal type to avoid this, but Decimal is slower and not used as often.)

```python
# 0.1 in binary is 0.000110011001100110011...
0.1 + 0.1 + 0.1
```

Behavior varies when you try to combine different types with operators.  Trying to use + to combine a string and an integer won't work.

```python
'Hello ' + 1111
```

The error message complained about not being able to "concatenate".  To concatenate means to "glue together", which is what + does when given two strings as operands.

```python
'Hello ' + 'world' + '!'
```

* We can change an int to a str (**cast** to str) with the str() function.  Then + will see two strings and concatenate them with no problem.
* Requiring an explicit cast like that helps prevent bugs where we didn't mean to concatenate.

```python
'Hello ' + str(1111)
```

## Review problem:  Degrees Fahrenheit

* We want to print "Temp: X F" where X is a temperature in Fahrenheit
* But we only know it's 20 degrees Celsius
* And the formula to convert is (C x 9/5) + 32
* Let's do this piece by piece, then all in one line

```python
# Step 1:  Calculate Celsius - let's just see what this value is...
20 * 9/5 + 32
```

```python
# Seems reasonable.  Step 2:  Print this value in the right format.
print('Temp: 68.0 F')
```

```python
# Step 3:  Can we do it all in one line?  Attempt 1 with a bug
print('Temp: ' , (20 * 9/5 + 32) , ' F') #fixed (turned + into ,)
```

```python
# Step 4: Oops, ints don't concatenate - need to cast to str
print('Temp: ' + str(20 * 9/5 + 32) + ' F') #or you could have done that too ig
```

# Summary of the High Points

* The Python interpreter executes one **statement** after another, from top to bottom.
* print() is an example of a **function** which takes **arguments** as input and behaves differently depending on those arguments.
    * They often return a useful value, but print() just returns None.
* **Expressions** are fragments of code that evaluate to something.  They can be nested inside each other.
* The four primitive types are **int**, **float**, **str**, and **bool**.
    * Float is used for decimal values but is slightly inaccurate.
