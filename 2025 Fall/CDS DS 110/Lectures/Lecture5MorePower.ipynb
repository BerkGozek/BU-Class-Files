# Boolean operators

We can make more expressive conditions for if/elif/else if we use boolean operators, which can build more complex conditions from simple ones.

*condition1* **and** *condition2* is True if and only if both condition1 and condition2 evaluate to True.

```python
percent = input('Enter a percentage between 0 and 100:')
if float(percent) >= 0 and float(percent) <= 100:
    if float(percent) >= 10:
        print('A decent return on investment....')
    else:
        print('Not a great return on investment....')
else:
    print('That is not in the requested range!')
```

*condition1* **or** *condition2* is True if and only if *at least one* of condition1 and condition2 is True.

```python
vip = False
spent = 10
if vip or spent >= 10000:
    print('Send this person a loyalty reward!')
else:
    print('This person deserves nothing!')
```

**not** *condition* is True if the condition is False, and False if the condition is True.

```python
vip = False
if not vip:
    print('Have you considered signing up to join the VIP program?')
else:
    print('Welcome back, VIP customer!')
```

The boolean operators can be strung together to create complex conditions.

While doing so, it's good to keep in mind that these have an **order of operations**, just like arithmetic.  All boolean operators come **after comparison**, but then **not comes first, then and, then or.**  Adding parentheses can change the order, just as in arithmetic.

```python
vip = False
spent = 0
if not vip or spent < 10000:  # "not" applied to vip before "or"
    print('Please spend more')
else:
    print('Hello, valued patron!')
```

```python
# This criterion doesn't make sense - it's just an example
# of changing the value using parens
vip = False
spent = 0
if not (vip or spent < 10000): # within parens evaluates to True
    print('Please spend more')
else:
    print('Hello, valued patron!')
```

It's tempting, because these are English words, to start being a little loose with the order of keywords in your code.  But not all English actually makes sense to the interpreter.  Think of them like + and *, which require specific operands (namely things that evaluate to True or False).

```python
my_list = [1,2,3]
my_list2 = [7,8,9]
if not 4 in my_list and not 4 in my_list2:
    print('No 4 found')
```

```python
my_list = [1,2,3]
my_list2 = [7,8,9]
# This works in English, but Python doesn't understand
# we're still talking about 4 in the second part
if 4 not in my_list and not in my_list2:
    print('This will actually cause an error - not how "in" works')
```

There generally isn't just one way to phrase a Boolean expression to do what you want.  There might be more multiple ways of doing the same thing.  For example, "De Morgan's Law" states that *not a and not b* is equvalent to *not (a or b)*. # Importing modules

Most functions and objects in Python are not immediately available to you unless you import the **module** that contains them.  A module is a text file that ends in .py (for Python) that contains definitions of functions, among other things.  Importing the module makes the functions available to use.

```python
import math
math.sqrt(2)
```

Calling the functions that belong to a module normally requires writing the module name first, followed by a dot, as if it were an object calling a method.  You can write something shorter if you give the module a nickname when you import it.

```python
import math as m
m.sqrt(2)
```

You can also import specfic functions from modules with "from...import...as".  If you do that, you don't need the module name in the call.

```python
from math import sqrt as my_sqrt
my_sqrt(2)
```

## Testing and debugging

Once you've written your program, what then?  **Programmers generally do not expect code to work the first time!**  Test your code, first with "normal" cases, and then with "weird" cases, also called "corner cases."

Does the following code work for ...

* A typical set of numbers?
* Negative numbers? (Corner case)
* Quitting the program immediately without errors? (Corner case)
* Entering non-number, non-done inputs? (Corner case)

```python
# From earlier while example
total = 0
count = 0
value_str = input('Enter a number, or "done" if done: ')
while value_str != 'done':
    count = count + 1
    value_int = int(value_str)
    total = total + value_int
    value_str = input('Enter a number, or "done" if done: ')
if count > 0:
    print('The average is ' + str(total/count))
```

The best way of fixing this last bug, using try/except, we won't cover until much later.  For the moment, we can decide we actually only cared about positive integers (and just make sure it doesn't error when given negative numbers).  isdigit() is a string method that returns true if a string is all digits 0-9, and we can check this to catch this bug.

```python
# Previous code, but we now check whether strings have digits
total = 0
count = 0
value_str = input('Enter a non-negative integer, or "done" if done: ')
while value_str != 'done':
    if not value_str.isdigit():
        print('Non-negative integers only!')
    else:
        count = count + 1
        value_int = int(value_str)
        total = total + value_int
    value_str = input('Enter a non-negative integer, or "done" if done: ')
if count > 0:
    print('The average is ' + str(total/count))
```

## Tracing and Using Debuggers

**Tracing** is just following along with code line-by-line and imagining what it is doing at each step.  Sometimes debugging just requires thinking carefully about what happens next.

Some programming environments have **debuggers** - features that allow you to step through your code line by line and inspect the values of variables as you go.  JupyterLab, a variant of Jupyter notebooks, offers a debugger (denoted with a little bug icon).  Tracing is like using your mind as a debugger.  Either approach can help you catch logical flaws in your code, but the actual debugger may be an improvement on your imagination when you're just starting out.

## Print debugging

Consider inserting temporary print statements that show what the values of variables are during the program.  This can give you more insight into what the program is doing.

```python
# Why is this program just sitting here?
# Let's print "value_str" because it controls the loop
total = 0
count = 0
value_str = input('Enter a number, or "done" if done: ')
while value_str != 'done':
    count = count + 1
    value_int = int(value_str)
    total = total + value_int
    print(value_str)
if count > 0:
    print('The average is ' + str(total/count))
```
