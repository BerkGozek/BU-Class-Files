# Problem 1: AI Memory (16 pts [4, 7, 5])

An AI chatbot needs to remember the last $k$ exchanges between a human and itself.  Each exchange consists of something a human said and something the AI said back, for example,


Human:  How are you?<br>
AI:  I am an AI and have no feelings.  Have a nice day!

(a, 4 pts) Define a simple Exchange class with a constructor `__init__(self, human, ai)` that initializes the two strings associated with the human and AI dialogue.  Override `__str__(self)` so that it returns output that looks like the example above ("Human:  " and "AI:  " added, and the two lines concatenated with a newline \n in-between).

```python
# TODO
class Exchange:
    def __init__(self,human,ai):
        self.human = human
        self.ai = ai

    def __str__(self):
        return f"Human: {self.human} \nAI: {self.ai}"
```

```python
print(Exchange("Hi","Hi!")) # expect "Human:  Hi\nAI:  Hi!"
```

(b, 7 pts) Define a `WindowedMemory` class.  The constructor should take a single argument *k* - this will define how many exchanges it can hold.  It should also initialize a list `memories` to be the empty list.  Then define a single method, `add_exchange(self, exchange)`.  If there are fewer than *k* exchanges in the memories list, add the exchange to the end of that list.  If there are *k* exchanges in the list already, drop the 0th element and then append to the end of the list.  Finally, override `__str__()` to concatenate all the exchanges, ending each with a newline.

```python
# TODO

class WindowedMemory:
    memory : list
    def __init__(self, k):
        self.memory = [""]*k

    def add_exchange(self,exchange):
        for i in range(len(self.memory)-1):
            self.memory[i+1]=str(self.memory[i])
            self.memory[i]= str(exchange)
    def __str__(self):
        retval : str = ""
        for i in self.memory:
            retval += i+"\n"
        return retval
```

# Problem 2:  The Book Class (42 points [2,3,3,3,4,7,8,7,5])

a) Create a Book class with the following methods:

* (2 pts) A constructor that takes the author (string), title (string), and a list of ratings (integers)
* (3 pts) A `__str__()` method that returns the string "[title] by [author]" when invoked
* (3 pts) An `average_rating()` method that takes no argument and returns the average of all the ratings in the list
* (3 pts) A `rate()` method that takes one integer argument and adds that integer to the end of the ratings, returning the new list of ratings

```python
# TODO Book class with __init__(), __str__(), average_rating(), rate()
class Book():
    author : str
    title : str
    ratings : list[int]

    def __init__(self,auth:str,tit:str,rat:list[int]):
        self.author = auth
        self.title = tit
        self.ratings = rat

    def __str__(self):
        return f"{self.title} by {self.author}"
    
    def average_rating(self):
        retval = 0
        for i in self.ratings:
            retval+= i
        return retval/len(self.ratings)
    
    def rate(self,nr:int):
        self.ratings.append(nr)
        return self.ratings
```

```python
# Tests
my_book = Book('Mary Shelley', 'Frankenstein', [5, 4, 3, 5, 3])
print(my_book) # expect 'Frankenstein by Mary Shelley'
print(my_book.average_rating()) # expect 4.0
print(my_book.rate(5)) # expect [5, 4, 3, 5, 3, 5]
```

c, 7 pts) Code a class that inherits from Book, Nonfiction.  Give it a constructor identical to Book's, except it has one additional argument:  the edition number (an int).  Then add to the Nonfiction class a method more_recent_than() which takes a Book as an argument and returns true if 1) the authors and titles match, *and* 2) the Book calling it has a higher edition number than the argument.

```python
# TODO Nonfiction class with __init__(), more_recent_than()
class Nonfiction(Book):
    edition : int
    def __init__(self, auth, tit, rat, edit):
        super().__init__(auth, tit, rat)
        self.edition = edit
    
    def more_recent_than(self, other : Nonfiction):
        return self.edition > other.edition
```

d, 8pts) Code a class Fiction that inherits from Book and has an extra attribute *sequels*, which should be a list of Fiction objects (or empty if there is no sequel).  Accept an extra argument in the constructor to set this attribute.  Throw a ValueError if any of the elements of the list are not Fiction objects.

Then write a method this_and_rest() that returns a string that lists the book itself followed by all the books in its sequels list.  The books should be turned to strings using the Book ``__str__()`` method, implicitly or explicitly.  Separate the books after the first with commas and spaces - see the example below.  (You may recall you can use "self" to refer to the object itself.)

```python
# TODO Fiction class with __init__(), this_and_rest()
from __future__ import annotations

class Fiction(Book):
    sequels : list[Fiction]

    def __init__(self, auth, tit, rat, seq : list[Fiction]):
        if not all(isinstance(s, Fiction) for s in seq):
            raise TypeError("All sequels must be Fiction Type")
        super().__init__(auth, tit, rat)
        self.sequels = seq
    
    def this_and_rest(self):
        retval = str(self) + ", "
        for i in self.sequels:
            retval+= str(i) + ", "
        return retval
```

# Problem 3:  Recursion (12 points [4,4,4])

a, 4 pts) Suppose some code takes a number of steps f(N) for input size N equal to f(N-1) + f(N-3) + 1.  f(0), f(1),  and f(2) are all 100, making f(3) = f(2) + f(0) + 1 = 201.  Write a function `running_time(N)` that recursively computes the number of steps required for input size N.

```python
# TODO

def running_time(i:int):
    if i <= 2:
        return 100
    return running_time(i-1) + running_time(i-3) + 1
```

```python
print(running_time(3)) # Expect 201
print(running_time(10)) # Expect 2827
```

b, 4 pts) Write a recursive function `recursive_lg()` that computes the log base 2 of the input, rounded up.  You can compute this by counting how many divisions by 2 you need to do to get a number that is 1 or less; but you need to do this recursively.

```python
# TODO
import math
def recursive_lg(i:int):
    if i==1:
        return 0
    elif i<2:
        return math.ceil(i/2)
    return 1 + recursive_lg(i/2)
```

```python
# Tests
print(recursive_lg(1)) # Expect 0
print(recursive_lg(4)) # Expect 2
print(recursive_lg(6)) # Expect 3
print(recursive_lg(1025)) # Expect 11
```
