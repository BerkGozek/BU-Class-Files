# Variables

```python
two_to_the_eighth = 2 ** 8
print(two_to_the_eighth)
```

<img src=" ">

* A **variable** stores the value **assigned** to it.
    * Above, a variable named two_to_the_eighth stores the value computed by the expression 2 ** 8.  This is an **assignment statement**.
    * Evaluating two_to_the_eighth in the print() produces 256, the value that was computed before.
    * The value is stored, not the expression.  2 ** 8 is not recomputed for the print statement.

The value stored is even remembered across code boxes.

```python
two_to_the_eighth * 2
```

Assigning a new value "clobbers" the value that was there before - that value is gone, replaced by the new value.

```python
pay_per_hour = 18
pay_per_hour = 20  # Pay raise!
print(pay_per_hour)
```

The expression on the right hand side of the equals is completely evaluated before it is assigned to the variable on the left hand side.  It's okay for the variable to appear on the right hand side - its new value may depend on itself.

```python
counter = 0
counter = counter + 1 # It's an instruction, not an equality!
print(counter)
```

```python
# We remember the value from one run to the next!
counter = counter + 1
print(counter)
```

How does one create a new variable?  Just introduce a new word (or **identifier**) on the left hand side of an assignment.

* You should avoid giving it the same name as an existing variable, keyword, or function.
* Try for short but self-explanatory names.
* For multiple words, link_with_underscores ("snake case").
* It must start with a letter or underscore.

```python
pay_per_hour = 20
hours = 40
total_pay = pay_per_hour * hours
print(total_pay)
```

Variable names, like much of Python, are *case-sensitive*:  if the capitalization is different, it's treated like a different variable.  Keeping a consistent name style helps to avoid the bugs that come from using a slightly off name.

```python
Pay_Per_Hour = 15   # please avoid this capitalization style!
print(pay_per_hour) # remembers the lowercase value
```

# Comparisons

Combined with conditional branching, comparisons allow a program to react differently to different inputs. We'll cover comparisons first.

Comparisons are expressions that evaluate to True or False.  One of the most common comparison operators is ==, which checks whether the left and right sides are equal.

```python
city = input('What city are we in? ')
print(city == 'Boston')
    
```

Equality generally cares about type, although the int 4 and the float 4.0 are considered equal.

```python
answer = input('What is 2+2? ')
print(answer == 4) # not going to work
```

```python
answer == '4'   # but this works
```

```python
int(answer) == 4 # or this
```

```python
float(answer) == 4 # or even this
```

Other comparisons include <, >, <= (less than or equal), >= (greater than or equal), and != (not equal).

* For numbers, these do what you would expect (except for occasional floating point error).
* Strings are compared in an order that is alphabetical if the strings are all lowercase letters; their order is complicated with more characters.

```python
print(1 < 1)
print(1 > 1)
print(1 != 1)
print(1 <= 1)
print(1 >= 1)
```

```python
print('aardvark' < 'zebra')
```

```python
print('capitalized' == 'Capitalized')
```

Comparison operators have a precedence just like addition and multiplication.  They happen after all the arithmetic.

```python
2 + 5 > 7 - 4  # 5 > 7 would be false, but (2+5) > (7-4) is True
```

# Conditional branching

We want our programs to be able to react differently to different inputs.  "If the user entered a negative value, complain.  Otherwise, add it to our running total..."



```python
total = 0
value_str = input('Enter a value: ')
value_int = int(value_str)
if value_int < 0:
    print('Sorry, that was a negative value.')
else:
    total = total + value_int
print(total)
```

```python
# not actually executable
if condition:
    statement_if_true1
    statement_if_true2
    statement_if_true3
    ...
else:
    statement_if_false1
    statement_if_false2
    ...
statement_regardless1
statement_regardless2
...
```

* On the same line as the "if" is a **condition** - an expression that evaluates to True or False, like a comparison.  (And then a : (colon).)
* The indented lines below that are the **if suite** - statements that execute sequentially only if the if's condition evaluated to True.
* The next line is "else:", followed by the indented **else suite** containing statements that execute one at a time if the condition was False.
* The de-indented code following that suite continues the program, happening regardless of which branch was taken.

```python
value = int(input('Enter an integer:'))
if value < 0:
    print('Negative')
else:
    print('Positive')
print('Done')
```

```python
password = input('Enter the password: ')
if password == '1234':
    print('Correct!')
    print('Your account has $1000000 in it.')
else:
    print('Incorrect.')
print('Have a nice day.')
```

```python
num1_str = input('Enter an integer: ')
num2_str = input('Enter a different integer: ')
num1_int = int(num1_str)
num2_int = int(num2_str)
if num1_str == num2_str:
    print('The numbers were supposed to be different...')
    print('But you entered ' + num1_str + ' twice!')
else:
    print(num2_str + ' divided by ' + num1_str + ' is...')
    print(num2_int / num1_int)  # Divide by zero would be error, btw
print('Done...')
```

"Else" and its suite are actually optional.  If you just have a lone "if", its suite happens if the condition is true, and regardless, the execution continues after the suite.

```python
language = input('What is your favorite language? ')
if language == 'Python':
    print('Mine too!')
print('But there sure are a lot of languages out there....')
```

"If" is a "two-way" branch, in that either the "if" suite or the "else" suite happens.  You may instead want a "multi-way" suite where:

    * if condition 1 is true, suite of code 1 happens
    * otherwise, if condition 2 is true, suite 2 happens
    * otherwise, if condition 3 is true, suite 3 happens
    * and so on; if no condition was true, one last suite happens
    
This is what if/elif/else does.

```python
value = int(input('Enter an integer between 0 and 100: '))
if value < 0:
    print('No negative numbers!')
elif value > 100:
    print('That value is too large!')
elif value == 42:
    print('That was the number I was thinking of!')
else:
    print('Guess again.')
    
```

<img src=" ">

An elif only triggers if all the previous if and elif conditions evaluated to False (and its condition evaluates to True).  Elif is short for "else if" which is short for putting another if in the "else" suite of the preceding "if."

```python
# Illustrating that the > 50 condition doesn't trigger if > 100 does
value = int(input('Enter an integer between 0 and 100: '))
if value < 0:
    print('No negative numbers!')
elif value > 100:
    print('That value is too large!')
elif value >= 50:
    print('Big!')
else:
    print('Small!')
```

```python
# Previous example is equivalent to the less stylish:
value = int(input('Enter an integer between 0 and 100: '))
if value < 0:
    print('No negative numbers!')
else:
    if value > 100:
        print('That value is too large!')
    else:
        if value >= 50:
            print('Big!')
        else:
            print('Small!')
```

Notice that we can put if's within other if or else suites.  If we would indent but have already indented, we add another tab to the indent level.

When a suite is executed because of an "if" or "else", it's executed just like any other code, including following the rules for "if" and "else."

In the next example, the fact that we want to comment on being an adult to all individuals 18 or over suggests breaking the inputs into two categories with a top-level "if" - adults and not-adults.  The two cases can then have more fine-grained distinctions with additional nested if and else suites.

```python
age = int(input('Enter your age: '))
if age < 18:
    if age < 5:
        print('Just a toddler, then.')
    elif age < 12:
        print('Not quite a teenager, then.')
    else:
        print('Teenage years ... a difficult time!')
else:
    print('An adult, then.')
    if age >= 55:
        print('And a senior citizen, too!')
```

## Review problem:  Min, max, mean, repeats

We can get three numbers from the user, and report which is the minimum, which is the max, and which is the mean (average).  We'll then also report on whether there were repeated numbers.

```python
num1 = int(input('First number: '))
num2 = int(input('Second number: '))
num3 = int(input('Third number: '))
my_max = max(num1, num2, num3)
my_min = min(num1, num2, num3)
my_mean = (num1+num2+num3)/3  # Note importance of parens!
print('Min: ' + str(my_min))
print('Max: ' + str(my_max))
print('Mean: ' + str(my_mean))
if num1 == num2:
    print(str(num1) + ' was repeated')
elif num2 == num3:
    print(str(num2) + ' was repeated')
elif num1 == num3:
    print(str(num3) + ' was repeated')
else:
    print('The numbers were unique')
# Note:  if all 3 are the same, is that handled correctly?
```
