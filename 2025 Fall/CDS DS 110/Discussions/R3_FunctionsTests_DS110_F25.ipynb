```python
# TODO
def big_sum_only(a:list[float],b:float):
    sum = 0
    for i in a:
        if i >= b:
            sum+=i
    return sum
```

5.  Write a function weave() that interleaves its two list arguments, so every even (starting from zero) element is from the first list and every odd element is from the second list.  You can assume the lists are of equal length.

```python
# TODO

def weave(a:list,b:list):
    retList = []
    for i in range(len(a)):
        retList.append(a[i])
        retList.append(b[i])
    return retList
```

6.  Write a function same_items() that takes two list arguments and is true iff every item in A can be found in B and every item in B can be found in A.

```python
# TODO

def same_items(a:list, b:list):

    for i in a:
        inList = False
        for ii in b :
            if i==ii:
                inList = True
                break
        if inList == False:
            return False
        
    for i in a:
        inList = False
        for ii in b :
            if i==ii:
                inList = True
                break
        if inList == False:
            return False
    return True

same_items([1,2,3],[2,3,1])
```

# Tool focus:  doctest

A classy way to write tests is to use doctest, which uses test cases that you provide
in comments with the function.  (You aren't expected to use this for homework or exams.)  Put the test examples in the "docstring" within the function, as shown here, then run that cell with the docstring inside, and then call doctest.testmod() to check that the tests work as expected.  (If no error is reported, it's fine.  You can try messing up the test and see what happens in that case.)

```python
import doctest
import statistics

def min_mean_max(lst):
    """
    Return the min, mean, and max value in lst as 3 return values.
    If the list is empty, return (0, 0, 0) instead.
    
    >>> min_mean_max([1,2,3,4,5])
    (1, 3, 5)
    
    >>> min_mean_max([])
    (0, 0, 0)
    """
    if not lst:  # meaning it's empty - some Python slickness
        return (0,0,0)
    return min(lst), statistics.mean(lst), max(lst)
```

# Tool Focus:  ChatGPT and GPT-4o

ChatGPT and GPT-4o are two "large language models," very recent AI models that have been trained on very large datasets of language to produce reasonable answers to questions.  ChatGPT is the most famous large language model, but GPT-4o is currently OpenAI's most advanced free model.  In the first phase of training, the models learn to predict what word comes next in a sequence of words; by feeding the predicted word back into the system, a next word can be generated, and so on to produce a whole reply to a "prompt."  In the next phase, the model adjusts itself in response to human feedback, so that the AI is overall more helpful than if it were just imitating what it had seen in its training data.

Note that this training process does not, by any means, guarantee a correct or helpful answer to a user's query.  The first phase of training just guarantees that likely-sounding sentences are generated, and the second just guarantees that helpful-sounding sentences are generated (except for harmful queries, which are suppressed in that phase of training).  Logic doesn't really figure into it (although other paid models try to make this reasoning go more carefully).  An analogy might be made with generative AI for images - if you ask for a mathematical diagram, like a picture of a normal distribution, you may get something that looks "mathy," but is not the requested diagram.

Despite those caveats, ChatGPT and similar models can potentially be very helpful to a programmer.  It can generate small programs in their entirety, although they need to be checked rigorously for bugs or mismatches with the user's intention.  It can write a suite of tests very quickly.  It can explain what some code does.  It can write documentation.  There are plenty of tasks that ChatGPT can help with, as long as its output is rigorously checked for errors.
