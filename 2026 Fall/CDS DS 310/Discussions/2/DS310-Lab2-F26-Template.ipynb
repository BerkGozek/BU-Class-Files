### Sample: writing relational algebra in a markdown cell

*(This is just a formatting reference — the query below is unrelated to any problem in this lab.)*

In a markdown cell, wrap relational-algebra expressions in `$...$` (inline) or `$$...$$` (own line) so Jupyter renders them with LaTeX/MathJax. Here are the symbols you'll use most often:

| Symbol | Meaning | LaTeX |
|---|---|---|
| $\pi$ | projection | `\pi` |
| $\sigma$ | selection | `\sigma` |
| $\Join$ | natural join | `\Join` |
| $\times$ | Cartesian product | `\times` |
| $\cup$ | union | `\cup` |
| $\cap$ | intersection | `\cap` |
| $-$ | set difference | `-` |
| $\rho$ | rename | `\rho` |

For example, given a toy schema `Employee(id, name, dept)` and `Department(dept, manager)`, the relational-algebra expression for "the names of employees in departments managed by Grace Hopper" is:

$$\pi_{name}(\sigma_{manager\ =\ 'Grace\ Hopper'}(Employee \Join Department))$$

which you'd type in the markdown cell as:

```
$$\pi_{name}(\sigma_{manager\ =\ 'Grace\ Hopper'}(Employee \Join Department))$$
```

Use spaces escaped as `\ ` and underscores in column names escaped as `\_` inside the subscripts, as shown above.
