# Problem 5: Find the min, max, and median (14 points)

This next program should prompt the user with "Enter a positive integer (-1 to quit)", getting a new number every time until the user enters -1. You can assume the input is an integer, but if the input is 0 or less, ignore it. After the program quits, print the min, max, and median values of the numbers entered on a single line. For example, if the input were 5, then 3, then 1, then 7, the output would be "min: 1, max: 7, median: 4". You may use the Python min and max functions. You must compute the median without the help of an outside module.

```python
#TODO
inp=int(input("Enter a positive integer (-1 to quit)"))
nums = []
while(inp!=-1):
    if(inp>0):
        nums.append(int(inp))
    inp=int(input("Enter a positive integer (-1 to quit)"))
nums.sort()

minimum = int(min(nums))
maximum = max(nums)

if(len(nums)%2==0):
    median = (nums[len(nums)//2]+nums[1+(len(nums)//2)])/2
else:
    median = nums[(len(nums)//2)]
print(f"min: {minimum}, max: {maximum}, median: {median}")
```
