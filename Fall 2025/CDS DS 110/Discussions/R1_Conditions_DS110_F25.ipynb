b) Similarly, write a program that asks how old the user is.  You can assume the input is an integer (although it will technically be a string until you convert the value with int()).

* If the number is at least 70, the program should reply, 'Wow!  You look good!'

* Otherwise, if the number is at least 18, reply 'So you're an adult!'

* Otherwise, if the number is greater than 0, reply 'So young...'   

* And last, if the number is 0 or less, reply "That's impossible!"

Remember that if a string contains an apostrophe, you should surround it with double quotes instead of single quotes.

```python
# TODO
age = int(input("How old are you?"))
if(age>=70):
    print("Wow!  You look good!")
elif(age>=18):
    print("So you're an adult!")
elif(age>=0):
    print("So young...")
else:
    print("That's impossible!")
```

## Problem 4:  Nested branching and computation

Take your age-asking program from the previous problem, and modify its behavior:

* If the age is over 70, after commenting that the user looks good, ask whether the user has grandkids.  If the answer is "yes", reply "Adorable!"  Otherwise, answer "Well, maybe someday...."
* If the age is between 0 and 18, reply "Just X years until you're 18!" where X is the difference between the age and 18 (e.g., if the input is 16, say "Just 2 years until you're 18!").

```python
# TODO
age0 = int(input("How old are you?"))
if (age0>70):
    if(input("Do you have grandchildren")=="yes"):
        print("Adorable!")
    else:
        print("Well, maybe someday...")
elif(age0>0 and age0<18):
    print(f"Just {18-age0} years until you're 18!")
```

## Problem 5:  Real estate

Suppose that the average value of an apartment in Somerville is about \$777 per square foot.

Write a program in the code box below that does the following:

* Ask for a number of square feet.
* Ask for an apartment price.
* Print the price per square foot.
* If the input results in a price less than \\$777 per square foot, print "It's a deal!"  If the input results in a price more than \\$777 per square foot, print "Overpriced!"  If the price is exactly \\$777 per square foot, print "Fair."

This is an extremely rough model of apartment prices - consider looking up a few apartments on Zillow and making the model more complicated if you have time.

```python
# TODO
sqrft = int(input("How large is the apartment you are selling?"))
price = int(input("How much are you pricing this apartment?"))
avg=777
cur=price/sqrft
print(str(cur)+"$/sqrft")
if(cur>avg):
    print("Overpriced!")
elif(cur==avg):
    print("Fair")
else:
    print("It's a deal!")
```
