# Short sample problem 1:  Loop (5 min)

Write a function that returns the next power of 2 after the argument (or the argument, if it is a power of 2).  For example, next_power(5) == 8, and next_power(8) == 8 as well.  You can assume the argument is at least 1.

```python
def next_power(input):
  p = 0

  while 2 ** p < input:
    p += 1

  return 2 ** p

print(next_power(5)) # Expect 8
print(next_power(8)) # Expect 8
print(next_power(32)) # Expect 32
print(next_power(33)) # Expect 64
```

# Short sample problem 2:  Dictionaries (5min)

Write a function that takes a list of strings and a dictionary as input.  Return a tuple, where the first element is the number of strings in the argument that are keys in the dictionary, and the second element is a list of the found keys' values.  For example, if test_dict = {
    "foo" : 1,
    "bar" : 3,
    "qux" : 2
} then find_strings(["foo", "bar", "baz"], test_dict) should return (2, [1,3]).

```python
def find_strings(strings, mydict):
  found_count = 0
  found_vals = []
  for string in strings:
    if string in mydict:
      found_count += 1
      found_vals.append(mydict[string])
  return found_count, found_vals

test_dict = {
    "foo" : 1,
    "bar" : 3,
    "qux" : 2
}
find_strings(["foo", "bar", "baz"], test_dict)
```

# Challenging short sample problem 3:  Nested loops (12 min)

Let's suppose some data has been corrupted in our 2D array of data, and we want to remove all rows that contain values less than 0.  Write a function good_rows() that takes this $m \times n$ 2D matrix of ints as an argument and returns an $(m - b) \times n$ 2D array that is missing the $b$ bad rows with negative values.

Hints:

* You might use the .shape attribute to iterate over the row and column numbers; recall it gives the number of rows and number of columns together as a tuple.

* Consider using a boolean to track whether the current row is good; you can reset it when you move to the next row.

* You can build a list of good rows, then covert back to a 2D array by passing the list to np.array().

```python
import numpy as np

def good_rows(maybe_bad):
    out = []
    for row in range(maybe_bad.shape[0]):
        keep = True
        for col in range(maybe_bad.shape[1]):
            if maybe_bad[row][col] < 0:
                keep = False
        if keep:
            out.append(maybe_bad[row,:])
    return np.array(out)

A = np.array([[1, 2, 3], [0, -1, 0], [4, 2, 2]])
good_rows(A)  # Expect array([[1,2,3],[4,2,2]])
```
